```python
# 安裝 OSMnx 與相關地理空間處理工具

import osmnx as ox
import networkx as nx
import numpy as np
import matplotlib.pyplot as plt
import time
import random
```

# Algorithm Core`n

```python
# --- 1. 基礎工具函式 ---
def generate_random_points(n, x_limit=100, y_limit=100):
    """在 100x100 的平面上產生 n 個隨機點"""
    return np.random.rand(n, 2) * [x_limit, y_limit]

def euclidean_dist(p1, p2):
    """計算兩點間的歐幾里得距離"""
    return np.linalg.norm(p1 - p2)

def get_covering_radius(points, centers):
    """
    計算覆蓋半徑 r(S):
    針對每個點，找出它離最近中心的距離。
    這些距離中的「最大值」就是覆蓋半徑。
    """
    if not centers:
        return float('inf')

    # 將 centers 轉為 numpy array 以利計算
    centers_arr = np.array(centers)

    # 計算每個點到所有中心的距離矩陣
    # 使用廣播機制 (Broadcasting) 計算所有點對距離
    dists = np.linalg.norm(points[:, np.newaxis] - centers_arr, axis=2)

    # 每個點離它「最近」中心的距離
    min_dists = np.min(dists, axis=1)

    # 找出這些最短距離中的「最大值」 (Minimax)
    radius = np.max(min_dists)
    return radius

# --- 2. 核心演算法實作 ---

def random_k_center(points, k):
    """隨機選取 k 個中心 (Baseline)"""
    indices = np.random.choice(len(points), k, replace=False)
    return points[indices].tolist()

def gonzalez_k_center(points, k):
    """
    Gonzalez 2-Approximation 演算法 (Greedy)
    1. 隨機選一個點當第一個中心
    2. 迭代選取離「當前中心集合」最遠的點作為下一個中心
    """
    # 初始化
    centers = []
    n = len(points)

    # 1. 隨機選第一個中心
    first_center_idx = np.random.randint(0, n)
    centers.append(points[first_center_idx])

    # 紀錄每個點到「目前已選中心」的最短距離
    # 初始時，因為只有一個中心，距離就是到該點的距離
    # 為了加速，我們維護一個陣列紀錄每個點目前的 min_dist
    current_min_dists = np.linalg.norm(points - centers[0], axis=1)

    # 2. 找剩下的 k-1 個中心
    for _ in range(k - 1):
        # 找出距離最遠的那個點的索引 (argmax)
        farthest_point_idx = np.argmax(current_min_dists)
        new_center = points[farthest_point_idx]
        centers.append(new_center)

        # 更新所有點的 min_dist (因為多了一個新中心，距離可能會變短)
        dist_to_new_center = np.linalg.norm(points - new_center, axis=1)
        current_min_dists = np.minimum(current_min_dists, dist_to_new_center)

    return centers
```

```python
# --- 3. 測試與視覺化 ---

# 設定參數
N = 100   # 點的數量
K = 5     # 中心的數量

# 產生資料
points = generate_random_points(N)

# 執行演算法
start_time = time.time()
gonzalez_centers = gonzalez_k_center(points, K)
gonzalez_time = time.time() - start_time
gonzalez_radius = get_covering_radius(points, gonzalez_centers)

start_time = time.time()
random_centers = random_k_center(points, K)
random_time = time.time() - start_time
random_radius = get_covering_radius(points, random_centers)

# 輸出結果
print(f"--- 測試結果 (N={N}, K={K}) ---")
print(f"Gonzalez 演算法: 半徑 r = {gonzalez_radius:.2f}, 耗時 = {gonzalez_time:.6f} 秒")
print(f"Random 隨機策略: 半徑 r = {random_radius:.2f}, 耗時 = {random_time:.6f} 秒")
print(f"改善幅度: {(random_radius - gonzalez_radius)/random_radius * 100:.2f}%")

# 繪圖比較
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# 繪製 Gonzalez 結果
centers_arr = np.array(gonzalez_centers)
ax1.scatter(points[:, 0], points[:, 1], c='blue', alpha=0.5, label='Points')
ax1.scatter(centers_arr[:, 0], centers_arr[:, 1], c='red', s=100, marker='*', label='Centers')
# 畫出覆蓋圓
for center in centers_arr:
    circle = plt.Circle(center, gonzalez_radius, color='red', fill=False, linestyle='--', alpha=0.3)
    ax1.add_patch(circle)
ax1.set_title(f'Gonzalez Greedy (r={gonzalez_radius:.1f})')
ax1.legend()
ax1.set_aspect('equal')

# 繪製 Random 結果
centers_arr_rnd = np.array(random_centers)
ax2.scatter(points[:, 0], points[:, 1], c='blue', alpha=0.5, label='Points')
ax2.scatter(centers_arr_rnd[:, 0], centers_arr_rnd[:, 1], c='green', s=100, marker='x', label='Centers')
# 畫出覆蓋圓
for center in centers_arr_rnd:
    circle = plt.Circle(center, random_radius, color='green', fill=False, linestyle='--', alpha=0.3)
    ax2.add_patch(circle)
ax2.set_title(f'Random Strategy (r={random_radius:.1f})')
ax2.legend()
ax2.set_aspect('equal')

plt.show()
```

# 第五階段 - 成本效益分析 (Cost-Benefit Analysis)

```python
# --- 第五階段：成本效益分析 (Trade-off Analysis) ---

# 1. 設定實驗參數
k_values = [5, 10, 15, 20, 30, 40, 50, 60] # 嘗試不同的 RSU 數量
results_greedy = []
results_random = []

print("開始批次實驗 (這可能會花一點時間)...")

# 2. 迴圈執行實驗
for k in k_values:
    # (A) 執行 Gonzalez Greedy
    centers_g = gonzalez_k_center(real_points, k)
    r_g = get_covering_radius(real_points, centers_g)
    results_greedy.append(r_g)

    # (B) 執行 Random Strategy (為了公平，取 5 次平均)
    temp_r_random = []
    for _ in range(5):
        centers_r = random_k_center(real_points, k)
        temp_r_random.append(get_covering_radius(real_points, centers_r))
    results_random.append(np.mean(temp_r_random))

    print(f"完成 k={k}: Greedy r={r_g:.0f}m, Random r={np.mean(temp_r_random):.0f}m")

# 3. 繪製 Trade-off Curve (權衡曲線圖)
plt.figure(figsize=(10, 6))

# 畫出兩條線
plt.plot(k_values, results_greedy, marker='o', label='Gonzalez Greedy (Ours)', color='red', linewidth=2)
plt.plot(k_values, results_random, marker='x', label='Random Strategy (Baseline)', color='green', linestyle='--')

# 標示圖表細節
plt.title('Cost-Benefit Analysis: RSU Count vs. Coverage Radius', fontsize=14)
plt.xlabel('Number of RSUs (k) - Cost', fontsize=12)
plt.ylabel('Max Coverage Radius (meters) - Service Quality', fontsize=12)
plt.grid(True, which='both', linestyle='--', alpha=0.7)
plt.legend(fontsize=12)

# 標示改善幅度 (Gap)
for i, k in enumerate(k_values):
    # 只在某些點標示，避免太擠
    if k in [5, 20, 50]:
        gap = results_random[i] - results_greedy[i]
        plt.arrow(k, results_random[i], 0, -gap+50, head_width=1.5, color='blue', length_includes_head=True)
        plt.text(k+1, (results_random[i] + results_greedy[i])/2, f'Better\nby {gap:.0f}m', color='blue', fontsize=9)

plt.show()

# 4. (選用) 將數據存成 CSV 方便貼到報告
import pandas as pd
df = pd.DataFrame({
    'k': k_values,
    'Radius_Greedy': results_greedy,
    'Radius_Random': results_random
})
# 儲存到 Google Drive (如果已掛載) 或暫存區
csv_path = 'tradeoff_results.csv'
df.to_csv(csv_path, index=False)
print(f"實驗數據已儲存為 {csv_path}")
print(df)
```

# 第六階段：進階實作 - 網路最短路徑距離 (Network Distance)
在第六階段前是歐基里德距離；
在第六階段後(包含在第六階段)是網路距離

## 步驟 1：前處理與計算距離矩陣 (Pre-processing)

```python
# --- 第六階段：進階實作 - 網路最短路徑距離 (Network Distance) ---

print("正在進行路網連通性檢查...")
# 1. 確保圖是強連通的 (移除孤島節點，避免距離無限大)
# 只保留最大的那個相連區塊
# 直接使用 NetworkX 函式庫尋找最大的強連通分量
scc = list(nx.strongly_connected_components(G_proj))
if scc:
    largest_component_nodes = max(scc, key=len)
    G_connected = G_proj.subgraph(largest_component_nodes).copy()
else:
    # 如果沒有強連通分量 (例如空圖)，則建立一個空圖
    G_connected = nx.MultiDiGraph() # Create an empty graph

nodes_connected = list(G_connected.nodes())
N_conn = len(nodes_connected)

# 建立 節點 ID 到 矩陣索引 (0 ~ N-1) 的對照表
node_to_idx = {node_id: i for i, node_id in enumerate(nodes_connected)}
idx_to_node = {i: node_id for i, node_id in enumerate(nodes_connected)}

print(f"原始節點數: {len(G_proj.nodes())}, 連通節點數: {N_conn}")
print("正在計算全對最短路徑矩陣 (All-Pairs Shortest Paths)... 這可能需要 1-2 分鐘...")

# 2. 計算距離矩陣 (Distance Matrix)
# 初始化一個 N x N 的矩陣，預設為無限大
dist_matrix = np.full((N_conn, N_conn), np.inf)

# 使用 NetworkX 的 Dijkstra 演算法計算所有路徑長度
# weight='length' 代表使用道路的物理長度 (公尺)
start_time = time.time()
path_gen = nx.all_pairs_dijkstra_path_length(G_connected, weight='length')

# 將結果填入矩陣
for source, target_dict in path_gen:
    i = node_to_idx[source]
    for target, length in target_dict.items():
        j = node_to_idx[target]
        dist_matrix[i, j] = length

calc_time = time.time() - start_time
print(f"距離矩陣計算完成！耗時: {calc_time:.2f} 秒")
print(f"矩陣大小: {dist_matrix.shape}")
```

## 步驟 2：修改演算法以支援矩陣輸入 (Algorithm Update)

```python
# --- 修改後的演算法：基於距離矩陣 (Matrix-based) ---

def get_covering_radius_matrix(dist_matrix, center_indices):
    """
    基於距離矩陣計算覆蓋半徑
    """
    if len(center_indices) == 0:
        return float('inf')

    # 將 center_indices 轉換為 numpy array 以處理索引
    center_indices_arr = np.array(center_indices)

    # 提取這些 center 對應的那些 row
    # dist_matrix[:, center_indices] 是一個 (N, k) 的矩陣
    # 代表每個點 (row) 到每個中心 (col) 的距離
    dists_to_centers = dist_matrix[:, center_indices_arr]

    # 每個點離最近中心的距離
    min_dists = np.min(dists_to_centers, axis=1)

    # 最大值即為覆蓋半徑
    return np.max(min_dists)

def gonzalez_k_center_matrix(dist_matrix, k):
    """
    針對距離矩陣優化的 Gonzalez 演算法
    回傳的是「矩陣索引 (indices)」，不是座標
    """
    n = dist_matrix.shape[0]
    centers = []

    # 1. 隨機選第一個中心 (索引)
    first_center = np.random.randint(0, n)
    centers.append(first_center)

    # 初始化 min_dists：所有點到第一個中心的距離
    current_min_dists = dist_matrix[:, first_center].copy()

    # 2. 找剩下的 k-1 個中心
    for _ in range(k - 1):
        # 找最遠的點
        farthest = np.argmax(current_min_dists)
        centers.append(farthest)

        # 更新距離：比較「原本的 min_dist」和「到新中心的距離」
        dist_to_new = dist_matrix[:, farthest]
        current_min_dists = np.minimum(current_min_dists, dist_to_new)

    return centers

def greedy_k_median_matrix(dist_matrix, k):
    """
    K-median 的貪婪演算法 (Greedy Approach)
    目標：最小化「所有點到最近中心的距離總和 (或平均)」
    邏輯：每次選擇能讓「總距離」下降最多的點作為新中心
    """
    n = dist_matrix.shape[0]
    centers = []

    # 1. 第一個點：為了簡化，選讓總距離最小的點 (1-median)
    # 計算每個點當中心時的總距離
    sum_dists = np.sum(dist_matrix, axis=0)
    first_center = np.argmin(sum_dists)
    centers.append(first_center)

    # 初始化 current_min_dists (每個點到目前中心的距離)
    current_min_dists = dist_matrix[:, first_center].copy()

    # 2. 找剩下的 k-1 個中心
    for _ in range(k - 1):
        # 這裡需要高效運算：
        # 我們要嘗試「如果選點 j 當新中心，新的總距離會是多少？」
        # 利用廣播 (Broadcasting) 一次算出所有候選點的效益

        # dist_matrix 的每一行 (col) 代表一個候選點到所有點的距離
        # 我們將 current_min_dists 與 dist_matrix 進行 element-wise minimum
        # 形狀變化: (N, 1) vs (N, N) -> (N, N)
        new_min_dists = np.minimum(current_min_dists[:, np.newaxis], dist_matrix)

        # 計算每個候選點產生的新總距離 (對 axis=0 加總)
        new_total_dists = np.sum(new_min_dists, axis=0)

        # 找出讓總距離最小的那個點
        best_candidate = np.argmin(new_total_dists)
        centers.append(best_candidate)

        # 更新 current_min_dists
        current_min_dists = np.minimum(current_min_dists, dist_matrix[:, best_candidate])

    return centers

def evaluate_metrics(dist_matrix, center_indices):
    """
    同時計算兩種指標：
    1. Max Radius (K-center 目標): 服務品質的最差情況 (越小越好)
    2. Average Distance (K-median 目標): 服務品質的平均情況 (越小越好)
    """
    if len(center_indices) == 0:
        return float('inf'), float('inf')

    # 將 center_indices 轉換為 numpy array 以處理索引
    center_indices_arr = np.array(center_indices)

    dists_to_centers = dist_matrix[:, center_indices_arr]
    min_dists = np.min(dists_to_centers, axis=1)

    max_radius = np.max(min_dists)
    avg_dist = np.mean(min_dists)

    return max_radius, avg_dist
```

## 步驟 3：執行實驗與視覺化 (Execution & Visualization)

```python
# --- 執行網路距離實驗 ---

K_net = 20  # RSU 數量

print(f"開始執行 Network Distance Gonzalez (k={K_net})...")
start_t = time.time()
# 注意：這裡輸入的是 dist_matrix
center_indices = gonzalez_k_center_matrix(dist_matrix, K_net)
exec_t = time.time() - start_t

# 計算半徑 (這是真實的道路行駛距離)
r_net = get_covering_radius_matrix(dist_matrix, center_indices)

print(f"--- 網路距離實驗結果 ---")
print(f"RSU 數量 (k): {K_net}")
print(f"最大覆蓋行駛距離 (r): {r_net:.2f} 公尺")
print(f"演算法執行時間: {exec_t:.6f} 秒")

# --- 視覺化 ---
# 將選出的索引轉換回 OSMnx 的節點 ID，再轉回座標以便繪圖
center_node_ids = [idx_to_node[idx] for idx in center_indices]
# 從 G_connected 中獲取這些節點的座標
x_vals = [G_connected.nodes[node]['x'] for node in center_node_ids]
y_vals = [G_connected.nodes[node]['y'] for node in center_node_ids]

# 繪圖
fig, ax = ox.plot_graph(G_connected,
              node_size=0,
              edge_color='#888888',
              edge_linewidth=0.5,
              show=False, close=False, figsize=(12, 12), bgcolor='white')

# 標示 RSU
ax.scatter(x_vals, y_vals, c='blue', s=200, marker='*', label='RSU (Network Dist)', zorder=5)

plt.title(f"Network Distance Optimization (k={K_net}, r={r_net:.0f}m)", fontsize=16)
plt.legend()
plt.show()
```

# 定義 Greedy K-median 演算法與評估工具

```python
# --- 第八階段：K-median 對比與網路距離終極實驗 ---

def greedy_k_median_matrix(dist_matrix, k):
    """
    K-median 的貪婪演算法 (Greedy Approach)
    目標：最小化「所有點到最近中心的距離總和 (或平均)」
    邏輯：每次選擇能讓「總距離」下降最多的點作為新中心
    """
    n = dist_matrix.shape[0]
    centers = []

    # 1. 第一個點：為了簡化，選讓總距離最小的點 (1-median)
    # 計算每個點當中心時的總距離
    sum_dists = np.sum(dist_matrix, axis=0)
    first_center = np.argmin(sum_dists)
    centers.append(first_center)

    # 初始化 current_min_dists (每個點到目前中心的距離)
    current_min_dists = dist_matrix[:, first_center].copy()

    # 2. 找剩下的 k-1 個中心
    for _ in range(k - 1):
        # 這裡需要高效運算：
        # 我們要嘗試「如果選點 j 當新中心，新的總距離會是多少？」
        # 利用廣播 (Broadcasting) 一次算出所有候選點的效益

        # dist_matrix 的每一行 (col) 代表一個候選點到所有點的距離
        # 我們將 current_min_dists 與 dist_matrix 進行 element-wise minimum
        # 形狀變化: (N, 1) vs (N, N) -> (N, N)
        new_min_dists = np.minimum(current_min_dists[:, np.newaxis], dist_matrix)

        # 計算每個候選點產生的新總距離 (對 axis=0 加總)
        new_total_dists = np.sum(new_min_dists, axis=0)

        # 找出讓總距離最小的那個點
        best_candidate = np.argmin(new_total_dists)
        centers.append(best_candidate)

        # 更新 current_min_dists
        current_min_dists = np.minimum(current_min_dists, dist_matrix[:, best_candidate])

    return centers

def evaluate_metrics(dist_matrix, center_indices):
    """
    同時計算兩種指標：
    1. Max Radius (K-center 目標): 服務品質的最差情況 (越小越好)
    2. Average Distance (K-median 目標): 服務品質的平均情況 (越小越好)
    """
    # [修改點] 使用 len() 檢查是否為空，解決 NumPy array 的 ValueError
    if len(center_indices) == 0:
        return float('inf'), float('inf')

    dists_to_centers = dist_matrix[:, center_indices]
    min_dists = np.min(dists_to_centers, axis=1)

    max_radius = np.max(min_dists)
    avg_dist = np.mean(min_dists)

    return max_radius, avg_dist
```

# 執行終極比較實驗 (Grand Experiment)

```python
# 設定 K 值
k_values_comp = [10, 20, 30, 40, 50, 60]

# 儲存結果
results = {
    'k': k_values_comp,
    'Gonzalez_MaxRadius': [], 'Gonzalez_AvgDist': [],
    'KMedian_MaxRadius': [],  'KMedian_AvgDist': [],
    'Random_MaxRadius': [],   'Random_AvgDist': []
}

print("開始執行終極比較實驗 (Gonzalez vs. K-median vs. Random)...")
print("使用指標：網路距離 (Network Distance)")

for k in k_values_comp:
    print(f"Running k={k}...")

    # 1. Gonzalez (K-center)
    centers_g = gonzalez_k_center_matrix(dist_matrix, k)
    r_g, avg_g = evaluate_metrics(dist_matrix, centers_g)
    results['Gonzalez_MaxRadius'].append(r_g)
    results['Gonzalez_AvgDist'].append(avg_g)

    # 2. Greedy K-median
    centers_m = greedy_k_median_matrix(dist_matrix, k)
    r_m, avg_m = evaluate_metrics(dist_matrix, centers_m)
    results['KMedian_MaxRadius'].append(r_m)
    results['KMedian_AvgDist'].append(avg_m)

    # 3. Random (取 5 次平均)
    temp_r, temp_avg = [], []
    for _ in range(5):
        # 隨機選 k 個索引
        rand_indices = np.random.choice(dist_matrix.shape[0], k, replace=False)
        r, avg = evaluate_metrics(dist_matrix, rand_indices)
        temp_r.append(r)
        temp_avg.append(avg)
    results['Random_MaxRadius'].append(np.mean(temp_r))
    results['Random_AvgDist'].append(np.mean(temp_avg))

# 轉為 DataFrame
df_comp = pd.DataFrame(results)
print("\n--- 實驗結果數據 ---")
print(df_comp[['k', 'Gonzalez_MaxRadius', 'KMedian_MaxRadius', 'Gonzalez_AvgDist', 'KMedian_AvgDist']])
```
